Using a list comprehesion and a variable called listOFInts, filter the items of the list down into only integers that are divisible by 7.

Ex: if the list had [-7,0,10,0,63,33] -> the filtered result would be [-7,0,63]


```python
listOfints=[-7,0,10,0,63,33]
listOfints=[num for num in listOfints if num%7==0]
listOfints 
```

*Q2*

Using a list comprehesion and a variable called listOfStrs,
save a new list called emails wich include any of the strings that had '@', 
while also removing all the spaces at the beginning  

Ex ["abc@123", "google.com", " goog@gmail.com"] 

```python
listOfStrs= ["abc@123", "google.com", " goog@gmail.com"] 
emails=[s.strip() for s in listOfStrs if "@" in s]
emails
```
